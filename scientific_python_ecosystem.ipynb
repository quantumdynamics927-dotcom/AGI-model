## 1. Numerical Computing with NumPy

**NumPy** is the backbone of the entire scientific Python ecosystem, providing:
- N-dimensional arrays for efficient numerical computation
- Linear algebra operations
- Random number generation
- Broadcasting capabilities

### Key Operations for Quantum Research

```python
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Golden ratio constant (used throughout AGI-model)
PHI = (1 + np.sqrt(5)) / 2
PHI_INV = 1 / PHI

print(f"Golden Ratio (φ): {PHI:.10f}")
print(f"Golden Ratio Inverse (φ⁻¹): {PHI_INV:.10f}")

# Create n-dimensional arrays (like VAE latent space)
latent_dim = 32
batch_size = 4
latent_vector = np.random.randn(batch_size, latent_dim)
print(f"\nLatent vector shape: {latent_vector.shape}")

# Linear algebra operations (density matrices, quantum states)
# Create a simple density matrix (Hermitian, trace=1)
psi = np.random.randn(4) + 1j * np.random.randn(4)
psi = psi / np.linalg.norm(psi)  # Normalize
density_matrix = np.outer(psi, np.conj(psi))

print(f"\nDensity matrix shape: {density_matrix.shape}")
print(f"Trace (should be 1): {np.trace(density_matrix):.10f}")
print(f"Is Hermitian: {np.allclose(density_matrix, density_matrix.T.conj())}")

# Phi-harmonic scaling (used in QNN agent)
phi_harmonic_sequence = np.array([PHI ** n for n in range(-3, 4)])
print(f"\nPhi-harmonic sequence: {phi_harmonic_sequence}")
```

## 2. Data Manipulation with Pandas

**Pandas** provides labeled data structures (DataFrames) for:
- Data cleaning and transformation
- Time-series analysis
- Statistical aggregations
- Data I/O operations

### Example: Consciousness Metrics DataFrame

```python
import pandas as pd
from datetime import datetime, timedelta

# Create a consciousness metrics DataFrame (simulating AGI training data)
dates = pd.date_range(start='2026-01-01', periods=10, freq='D')
n_samples = len(dates)

consciousness_data = pd.DataFrame({
    'timestamp': dates,
    'phi_alignment': np.random.uniform(0.5, 0.7, n_samples),
    'consciousness_level': np.random.uniform(0.3, 0.9, n_samples),
    'entanglement_entropy': np.random.uniform(0.1, 0.5, n_samples),
    'quantum_fidelity': np.random.uniform(0.8, 1.0, n_samples),
    'training_epoch': range(1, n_samples + 1)
})

# Set timestamp as index for time-series analysis
consciousness_data.set_index('timestamp', inplace=True)

print("Consciousness Metrics DataFrame:")
print(consciousness_data)

# Data transformation and statistics
print("\n--- Statistical Summary ---")
print(consciousness_data.describe())

# Time-series operations (rolling averages)
print("\n--- Rolling Average (3-day window) ---")
rolling_avg = consciousness_data['phi_alignment'].rolling(window=3).mean()
print(rolling_avg)

# Filtering based on consciousness threshold
high_consciousness = consciousness_data[consciousness_data['consciousness_level'] > 0.6]
print(f"\nHigh consciousness samples (>0.6): {len(high_consciousness)}")
```
